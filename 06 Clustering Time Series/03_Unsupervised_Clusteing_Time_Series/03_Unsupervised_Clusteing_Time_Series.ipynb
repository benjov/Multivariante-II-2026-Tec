### 2.2 Agrupación

Como siempre, generar, conseguir o procesar los datos es lo complicado, el modelo es algo así:

```python
#

data_array = np.array( time_series_data.T.drop( ['Periodo', 'Fecha', 'Hora'] ).values )

data_array
```

```python
#

model = TimeSeriesKMeans( n_clusters = 2, metric = "dtw", max_iter = 10, random_state = 2026, n_jobs = -1)
#metric{“euclidean”, “dtw”, “softdtw”} (default: “euclidean”)

model.fit(data_array)
```

```python
model.inertia_ #model.labels_
```

```python
#

clusters = []

#
# k de 2 a 6: con DTW cada ajuste es costoso (k de 1 a 10 tardaba más de 15 minutos)
for i in range(2, 7):
    km = TimeSeriesKMeans( n_clusters = i, metric = "dtw", max_iter = 10, random_state = 2026, n_jobs = -1 ).fit( data_array )
    clusters.append( km.inertia_ )
```

```python
# Gráfica del codo: inercia (suma de distancias DTW al centroide) contra k
ks = list(range(2, 7))
fig, ax = plt.subplots( figsize = (12, 8) )
sns.lineplot(x = ks, y = clusters, marker = 'o', ax = ax)
ax.set_title('Búsqueda del codo')
ax.set_xlabel('Número de grupos k')
ax.set_ylabel('Inercia')
ax.set_xticks(ks)
plt.show()
```

```python
# selección Final:

#

model = TimeSeriesKMeans( n_clusters = 3, metric = "dtw", max_iter = 10, random_state = 2026, n_jobs = -1)
#metric{“euclidean”, “dtw”, “softdtw”} (default: “euclidean”)

model.fit(data_array)
```

```python
#

Nodos = time_series_data.T.drop( ['Periodo', 'Fecha', 'Hora'] ).index.tolist()

Nodos
```

```python
# Prediciendo clases

y = model.predict(data_array)

y
```

```python
#

x = time_series_data.Periodo
```

```python
np.random.seed(2026)   # para que la selección aleatoria de nodos sea reproducible
plt.figure( figsize = (20,20) )

k_dict = {'1':0,'2':1,'3':2} # Diccionario de labels, corrected

colors = ['cyan', 'darkblue', 'darkred']

Names = ['Class 0', 'Class 1', 'Class 2']

for j in range(1,4):
    plt.subplot( 3, 3, j)

    k = np.random.choice( np.where( y == k_dict[ str(j) ] )[0] ) # Selección aleatoria de casos

    plt.plot( x, data_array[k], '.' ,color = colors[j-1] )
    plt.ylabel( 'PML-MDA', fontsize = 20 )
    plt.xlabel( 'Time', fontsize = 20)
    plt.title( 'Nodo=%s, Class = %s'%(Nodos[k], Names[j-1]), fontsize = 20)
    plt.ylim( data_array.min(), data_array.max() )
    plt.xticks(rotation = 90) # Rotate x-axis labels vertically
```

```python
# Datos Finales
data = {
    'Nodo': Nodos,
    'Grupo': y
}

df_Final = pd.DataFrame( data )

df_Final
```
